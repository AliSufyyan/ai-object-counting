```python
import cv2

image=cv2.imread('street.jpg')

if image is None:
    raise FileNotFoundError("Image not found")
```

```python
print('image Shape ',image.shape)
```

```python
img_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

```python
resized=cv2.resize(image,(640,480))
```

```python
cv2.rectangle(
    resized,
    (50,60),
    (250,300),
    (0,255,0),
    2
)
```

```python
cv2.putText(
    resized,
    "car",
    (50,45),
    cv2.FONT_HERSHEY_COMPLEX,
    0.8,
    (0,255,0),
    2
    )
```

```python
cv2.imwrite('annotated.jpg',resized)
```
